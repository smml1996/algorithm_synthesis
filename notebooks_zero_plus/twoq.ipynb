```python
import os, sys
sys.path.append(os.getcwd() + "/../")
import notebook_utils
from experiments_utils import TwoQZeroPlusExperimentID
```

```python
experiments = [TwoQZeroPlusExperimentID.HCXH2, TwoQZeroPlusExperimentID.HCXH, TwoQZeroPlusExperimentID.TWOQ, TwoQZeroPlusExperimentID.TWOQ2]
```

```python
from experiments_utils import load_all_lambdas
import pandas as pd

experiment_ids = []
hardware_specs = []
embedding_indices = []
horizons = []
lambdas = []
for experiment_id in experiments:
    expr_lambdas = load_all_lambdas(experiment_id)
    
    for (hardware_spec, embedding_dict) in expr_lambdas.items():
        for (embedding_index, horizon_dict) in embedding_dict.items():
            for (horizon, lam) in horizon_dict.items():
                experiment_ids.append(experiment_id.value)
                hardware_specs.append(hardware_spec.value)
                embedding_indices.append(embedding_index)
                horizons.append(horizon)
                lambdas.append(lam)
                
all_lambdas = pd.DataFrame.from_dict({
            'experiment': experiment_ids,
            'hardware_spec': hardware_specs,
            'embedding_index': embedding_indices,
            'horizon': horizons,
            'guarantee': lambdas
        }) 
all_lambdas
```

```python
df_pivoted = all_lambdas.pivot(
    index=['hardware_spec', 'embedding_index', 'horizon'],
    columns='experiment',
    values='guarantee'
).reset_index()
df_pivoted
```

```python
df_pivoted["h-h2"] = df_pivoted.hcxh - df_pivoted.hcxh2
df_pivoted["h-q"] = df_pivoted.hcxh - df_pivoted.twoq
df_pivoted["h-q2"] = df_pivoted.hcxh - df_pivoted.twoq2
df_pivoted["h2-q"] = df_pivoted.hcxh2 - df_pivoted.twoq
df_pivoted["h2-q2"] = df_pivoted.hcxh2 - df_pivoted.twoq2
df_pivoted["q-q2"] = df_pivoted.twoq - df_pivoted.twoq2
```

```python
df_pivoted.to_csv("all_lambdas.csv")
```
