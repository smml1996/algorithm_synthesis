```python
from qiskit import QuantumCircuit, transpile
from qiskit.circuit import Parameter
from qiskit.circuit.library import *
from math import sqrt
```

```python
theta = Parameter("theta")
phi = Parameter("phi")
lam = Parameter("lam")
gamma = Parameter("gamma")

gates = [AndGate(2), BitwiseXorGate(2, 1), C3SXGate(), C3XGate(), CCXGate(), CCZGate(), CHGate(), CPhaseGate(theta), CRXGate(theta), CRYGate(theta), CRZGate(theta), CSdgGate(), CSGate(), CSwapGate(), CSXGate(), CU1Gate(theta), CU3Gate(theta, phi, lam), CUGate(theta, phi, lam, gamma), CXGate(), CYGate(), CZGate(), DCXGate(), ECRGate(), FullAdderGate(2), HGate(), iSwapGate(), OrGate(2), PhaseGate(theta), RC3XGate(), RCCXGate(), RVGate(0.1, 0.2, 0.3), RXGate(theta), RXXGate(theta), RYGate(theta), RYYGate(theta), RZGate(phi), RZXGate(theta), RZZGate(theta), SdgGate(), SXGate(), TdgGate(), TGate(), U1Gate(theta), U2Gate(theta, phi), 
        #  U3Gate(theta, phi, lam), 
        YGate(), ZGate(), XGate()]

f"There are {len(gates)} gates"
```

```python

gates_dict = dict()
for gate in gates:
    num_decomposition = len(gate.decompositions)
    instructions_used = set()
    for decomposition in gate.decompositions:
        for instruction, qargs, cargs in decomposition.data:
            instruction_str = instruction.name + ",".join([str(x) for x in qargs]) + "--" +",".join([str(x) for x in cargs])
            instructions_used.add(instruction_str)
    if gate.name not in gates_dict.keys():
        if len(instructions_used) < 7 and len(instructions_used) > 0:
            gates_dict[gate.name] = (num_decomposition, instructions_used)
    else:
        (prev_decomp, prev_instructions_used) = gates_dict[gate.name]
        assert prev_decomp == num_decomposition
        assert instructions_used == prev_instructions_used
        
```

```python
sorted_keys = sorted([x for x in gates_dict.keys()], key = lambda x: (gates_dict[x][0], -len(gates_dict[x][1])), reverse=True)
for key in sorted_keys:
    print(key, gates_dict[key][0], len(gates_dict[key][1]))
```

```python
def print_decompositions(gate):
    for decomposition in gate.decompositions:
        print(decomposition)
```

```python
def get_decomposition_for_basis(gate, basis_gates, num_qubits=1, params=[1]):
    qc = QuantumCircuit(num_qubits)
    qc.append(gate, params)
    transpiled_qc = transpile(qc, basis_gates=basis_gates)
    print(transpiled_qc)
```
